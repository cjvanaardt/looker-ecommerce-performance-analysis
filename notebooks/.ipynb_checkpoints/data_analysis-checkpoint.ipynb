### How have retention and repeat purchase patterns changed over the last 24–36 months, segmented by acquisition channel, demographics (age/gender), and location?
Metrics of interest: Retention and repeat purchase

    - number of orders
    - average number of items per order
    - average price per order
    - average price per item

```python
user_orders = users.merge(order_items, left_on='id', right_on='user_id')
user_orders
```

```python
# repeated_orders
multiple_orders = user_orders.groupby('id_x').filter(lambda x: x['order_id'].nunique() > 1)

# want to only keep orders past the first one
multiple_orders = multiple_orders.sort_values(by=['id_x', 'order_id'])
multiple_orders['order_number'] = multiple_orders.groupby('id_x')['order_id'].rank(method='dense').astype(int)
repeated_orders = multiple_orders[multiple_orders['order_number'] > 1].copy()
```

```python
result = repeated_orders.groupby(repeated_orders['created_at_y'].dt.to_period('M')).agg(
    n_orders=('order_id', 'nunique'),
    n_items=('id_y', 'count'),
    tot_rev=('sale_price', 'sum')
)
result
```

```python
plt.plot(result.index.to_timestamp(), result['n_orders'])
```

```python
plt.plot(result.index.to_timestamp(), result['n_items'])
```

```python
plt.plot(result.index.to_timestamp(), result['n_items'] / result['n_orders'])
```

```python
plt.plot(result.index.to_timestamp(), result['tot_rev'] / result['n_orders'])
```

```python
plt.plot(result.index.to_timestamp(), result['tot_rev'] / result['n_items'])
```
